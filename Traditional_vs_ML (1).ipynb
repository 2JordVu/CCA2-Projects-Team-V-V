# Traditional Programming vs Machine Learning — Loan Approval
**Group assignment** · Input: `loan_input.xlsx` · Output: `loan_output.xlsx`

| | Traditional programming | Machine learning |
|---|---|---|
| What we write | The **rules** (if/else) | The **data** (past decisions) |
| What the computer does | Applies the rules to new applicants | **Learns** the rules from past decisions |
| Changing behaviour | Edit the code / thresholds | Re-train on new data |

**Keeping the files in sync:** edit any number in `loan_input.xlsx` (or a threshold on its Assumptions sheet), save, then run this notebook top to bottom (Runtime > Run all). `loan_output.xlsx` is rebuilt from the input every time, so every change, including the ML results, appears in it.

**Flow of this notebook**
1. Read 20 applicants from Excel (pandas + openpyxl)
2. Traditional approach: hand-written rules
3. ML approach: logistic regression learns from `Past_Decision`
4. Compare both approaches
5. Write results back to Excel (decision columns are live formulas, so edits to the numbers update the brackets/decisions)

```python
# Step 0: libraries (already available on Colab; run `pip install pandas openpyxl scikit-learn` elsewhere)
import os, subprocess, shutil
import numpy as np
import pandas as pd
import openpyxl
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import LeaveOneOut, cross_val_predict

INPUT_FILE  = "loan_input.xlsx"
OUTPUT_FILE = "loan_output.xlsx"

# On Google Colab: upload the input file if it is not already in the session
if not os.path.exists(INPUT_FILE):
    try:
        from google.colab import files
        print("Upload loan_input.xlsx ...")
        files.upload()
    except ImportError:
        raise FileNotFoundError(f"Put {INPUT_FILE} in the same folder as this notebook.")
print("Input file found:", os.path.abspath(INPUT_FILE))
```

## Step 1 — Read the input Excel file
The workbook has two sheets: **Applicants** (the data) and **Assumptions** (interest rate, credit bands, rule thresholds).
In Excel, columns J–O are formulas (monthly income, proposed EMI, DTI, loan-to-income, and the credit/DTI "brackets").
Here we recompute them in pandas so the notebook never depends on cached Excel values.

```python
raw = pd.read_excel(INPUT_FILE, sheet_name="Applicants")
raw = raw[raw["Applicant_ID"].astype(str).str.match(r"^APP\d+$")].copy()   # drop note rows

df = pd.DataFrame({
    "Applicant_ID": raw["Applicant_ID"],
    "Name": raw["Name"],
    "Age": raw["Age"].astype(int),
    "Annual_Income": raw["Annual_Income (INR)"].astype(float),
    "Credit_Score": raw["Credit_Score"].astype(int),
    "Employment_Years": raw["Employment_Years"].astype(int),
    "Existing_Monthly_EMI": raw["Existing_Monthly_EMI (INR)"].astype(float),
    "Loan_Amount": raw["Loan_Amount (INR)"].astype(float),
    "Loan_Term_Years": raw["Loan_Term_Years"].astype(int),
    "Past_Decision": raw["Past_Decision (label)"],
}).reset_index(drop=True)

# Read the thresholds from the Assumptions sheet so Excel and Python always agree
ws_a = openpyxl.load_workbook(INPUT_FILE)["Assumptions"]
P = {
    "rate": ws_a["B2"].value, "excellent": ws_a["B3"].value, "good": ws_a["B4"].value, "fair": ws_a["B5"].value,
    "dti_low": ws_a["B6"].value, "dti_mod": ws_a["B7"].value,
    "min_score": ws_a["B8"].value, "max_dti": ws_a["B9"].value, "min_emp": ws_a["B10"].value, "max_lti": ws_a["B11"].value,
}
print(P)

# Derived columns: same maths as the Excel formulas
def emi(loan, years, annual_rate):
    r = annual_rate / 12; n = years * 12
    return loan * r / (1 - (1 + r) ** (-n))

df["Monthly_Income"] = df["Annual_Income"] / 12
df["Proposed_EMI"]   = emi(df["Loan_Amount"], df["Loan_Term_Years"], P["rate"])
df["DTI"]            = (df["Existing_Monthly_EMI"] + df["Proposed_EMI"]) / df["Monthly_Income"]
df["Loan_to_Income"] = df["Loan_Amount"] / df["Annual_Income"]
df["Credit_Band"]    = np.select([df.Credit_Score >= P["excellent"], df.Credit_Score >= P["good"], df.Credit_Score >= P["fair"]],
                                 ["Excellent", "Good", "Fair"], default="Poor")
df["DTI_Band"]       = np.select([df.DTI <= P["dti_low"], df.DTI <= P["dti_mod"]], ["Low", "Moderate"], default="High")

# Sanity check against the values Excel has cached (only if the file was saved by Excel/LibreOffice)
cached = raw["DTI (total EMI / income)"].astype(float).reset_index(drop=True)
if cached.notna().all():
    assert np.allclose(cached, df["DTI"], atol=1e-6), "Python DTI differs from Excel DTI"
    print("Check passed: Python DTI matches the Excel formulas.")
print(f"{len(df)} applicants loaded")
df.head()
```

## Step 2 — Traditional programming: we write the rules
A loan officer's policy is turned directly into code. **Rules in, decision out.**

Approve only if **all** hold: credit score ≥ min, DTI ≤ max, employment ≥ min years, loan-to-income ≤ max
(thresholds come from the `Assumptions` sheet).

```python
def rule_based_decision(row, p=P):
    reasons = []
    if row["Credit_Score"] < p["min_score"]:       reasons.append("Low credit score")
    if row["DTI"] > p["max_dti"]:                  reasons.append("High DTI")
    if row["Employment_Years"] < p["min_emp"]:     reasons.append("Short employment")
    if row["Loan_to_Income"] > p["max_lti"]:       reasons.append("Loan too large vs income")
    return ("Approved", "All criteria met") if not reasons else ("Rejected", "; ".join(reasons))

df[["Rule_Decision", "Rule_Reason"]] = df.apply(lambda r: pd.Series(rule_based_decision(r)), axis=1)
df[["Applicant_ID", "Name", "Credit_Score", "DTI", "Employment_Years", "Rule_Decision", "Rule_Reason", "Past_Decision"]].round(3)
```

## Step 3 — Machine learning: the computer learns from past decisions
We do **not** write any rules here. We give the model the applicant numbers and the historical `Past_Decision`
(what loan officers actually decided) and let **logistic regression** find the pattern.

Only 20 rows is very small, so to be honest about accuracy we use **leave-one-out cross-validation**:
each applicant is predicted by a model trained on the *other 19*, so the model never sees the answer it is predicting.

```python
FEATURES = ["Age", "Annual_Income", "Credit_Score", "Employment_Years", "DTI", "Loan_to_Income"]
X = df[FEATURES].values
y = (df["Past_Decision"] == "Approved").astype(int).values

model = make_pipeline(StandardScaler(), LogisticRegression(C=1.0, max_iter=1000))

# Out-of-sample probability of approval for every applicant (leave-one-out)
df["ML_Probability"] = cross_val_predict(model, X, y, cv=LeaveOneOut(), method="predict_proba")[:, 1]
df["ML_Decision"]    = np.where(df["ML_Probability"] >= 0.5, "Approved", "Rejected")

# Final model trained on all 20 rows (used to inspect what it learned and for the what-if cell)
model.fit(X, y)
coef = pd.Series(model[-1].coef_[0], index=FEATURES).sort_values()
print("What the model learned (standardised coefficients; positive = pushes towards Approve):")
print(coef.round(2).to_string())
df[["Applicant_ID", "Name", "ML_Probability", "ML_Decision", "Past_Decision"]].round(3)
```

## Step 4 — Compare the two approaches

```python
df["Rule_vs_ML"]        = np.where(df["Rule_Decision"] == df["ML_Decision"], "Agree", "Disagree")
df["Rule_Correct"]      = np.where(df["Rule_Decision"] == df["Past_Decision"], "Yes", "No")
df["ML_Correct"]        = np.where(df["ML_Decision"]   == df["Past_Decision"], "Yes", "No")

n = len(df)
print(f"Rule-based approvals : {(df.Rule_Decision=='Approved').sum()} / {n}")
print(f"ML approvals         : {(df.ML_Decision=='Approved').sum()} / {n}")
print(f"Past (historical)    : {(df.Past_Decision=='Approved').sum()} / {n}")
print(f"Rules vs ML agree    : {(df.Rule_vs_ML=='Agree').sum()} / {n}")
print(f"Rule accuracy vs Past: {(df.Rule_Correct=='Yes').mean():.0%}")
print(f"ML accuracy vs Past  : {(df.ML_Correct=='Yes').mean():.0%}  (leave-one-out)")

print("\nApplicants where Rules and ML disagree:")
df.loc[df.Rule_vs_ML == "Disagree",
       ["Applicant_ID", "Name", "Credit_Score", "Employment_Years", "DTI", "Rule_Decision", "ML_Decision", "ML_Probability", "Past_Decision"]].round(3)
```

```python
# What-if: a brand-new applicant goes through both approaches
new = {"Age": 30, "Annual_Income": 900000, "Credit_Score": 700, "Employment_Years": 4,
       "Existing_Monthly_EMI": 8000, "Loan_Amount": 1500000, "Loan_Term_Years": 10}
n_emi = emi(new["Loan_Amount"], new["Loan_Term_Years"], P["rate"])
new["DTI"] = (new["Existing_Monthly_EMI"] + n_emi) / (new["Annual_Income"] / 12)
new["Loan_to_Income"] = new["Loan_Amount"] / new["Annual_Income"]

print("Traditional :", rule_based_decision(new))
p = model.predict_proba(np.array([[new[f] for f in FEATURES]]))[0, 1]
print(f"ML          : {'Approved' if p >= 0.5 else 'Rejected'}  (probability of approval = {p:.2f})")
```

## Step 5 — Write the results back to Excel
The output workbook is the input workbook plus new columns:
* **Rule_Decision / Rule_Reason / Rule_vs_ML / correctness columns / Summary sheet are Excel formulas** — edit any applicant number or any threshold on the `Assumptions` sheet and they update instantly.
* **ML_Probability is a stored value** produced by the trained model (a model cannot live inside a cell). `ML_Decision` is a formula on top of it. After changing the data, re-run this notebook to re-train the model.

```python
wb = openpyxl.load_workbook(INPUT_FILE)          # keeps the input formulas intact
ws = wb["Applicants"]

hdr_fill = PatternFill("solid", fgColor="375623"); ml_fill = PatternFill("solid", fgColor="7030A0")
thin = Side(style="thin", color="BFBFBF"); bd = Border(left=thin, right=thin, top=thin, bottom=thin)
new_heads = ["Rule_Decision", "Rule_Reason", "ML_Probability_Approve", "ML_Decision", "Rule_vs_ML", "Rule_Matches_Past", "ML_Matches_Past"]
for j, h in enumerate(new_heads):
    c = ws.cell(1, 17 + j, h)
    c.font = Font(bold=True, color="FFFFFF"); c.alignment = Alignment(wrap_text=True, horizontal="center", vertical="center")
    c.fill = ml_fill if h.startswith("ML_Prob") else hdr_fill

A = "Assumptions!"
for i in range(len(df)):
    r = i + 2
    ws.cell(r, 17, f'=IF(AND(E{r}>={A}$B$8,L{r}<={A}$B$9,F{r}>={A}$B$10,M{r}<={A}$B$11),"Approved","Rejected")')
    ws.cell(r, 18, f'=IF(Q{r}="Approved","All criteria met",SUBSTITUTE(TRIM(IF(E{r}<{A}$B$8,"Low credit score; ","")&IF(L{r}>{A}$B$9,"High DTI; ","")'
                   f'&IF(F{r}<{A}$B$10,"Short employment; ","")&IF(M{r}>{A}$B$11,"Loan too large vs income; ",""))&"#END",";#END",""))')
    ws.cell(r, 19, round(float(df.loc[i, "ML_Probability"]), 4))                       # value from the ML model
    ws.cell(r, 20, f'=IF(S{r}>=0.5,"Approved","Rejected")')
    ws.cell(r, 21, f'=IF(Q{r}=T{r},"Agree","Disagree")')
    ws.cell(r, 22, f'=IF(Q{r}=P{r},"Yes","No")')
    ws.cell(r, 23, f'=IF(T{r}=P{r},"Yes","No")')
    ws.cell(r, 19).number_format = "0.0%"
    for c in range(17, 24): ws.cell(r, c).border = bd
for col, w in zip("QRSTUVW", [14, 34, 14, 13, 12, 14, 13]): ws.column_dimensions[col].width = w
ws["A23"] = ("Blue font = editable inputs. Columns J:O and Q:W (except S) are formulas and update automatically. "
             "Column S (ML probability) comes from the notebook's trained model - re-run the notebook to refresh it.")

# Summary sheet (all formulas)
sm = wb.create_sheet("Summary", 1)   # Applicants stays the first sheet
rows = [
 ("Metric", "Value"),
 ("Applicants", "=COUNTA(Applicants!A2:A21)"),
 ("Approved - traditional rules", '=COUNTIF(Applicants!Q2:Q21,"Approved")'),
 ("Approved - machine learning", '=COUNTIF(Applicants!T2:T21,"Approved")'),
 ("Approved - historical (Past_Decision)", '=COUNTIF(Applicants!P2:P21,"Approved")'),
 ("Rules and ML agree (count)", '=COUNTIF(Applicants!U2:U21,"Agree")'),
 ("Rules and ML disagree (count)", '=COUNTIF(Applicants!U2:U21,"Disagree")'),
 ("Rule accuracy vs Past_Decision", '=COUNTIF(Applicants!V2:V21,"Yes")/B2'),
 ("ML accuracy vs Past_Decision (leave-one-out)", '=COUNTIF(Applicants!W2:W21,"Yes")/B2'),
]
for row in rows: sm.append(row)
for c in sm[1]: c.font = Font(bold=True, color="FFFFFF"); c.fill = PatternFill("solid", fgColor="1F3864")
sm["B8"].number_format = "0%"; sm["B9"].number_format = "0%"
sm.column_dimensions["A"].width = 46; sm.column_dimensions["B"].width = 12
notes = ["", "How to read this file",
 "- Traditional: rules written by a programmer (thresholds on the Assumptions sheet).",
 "- ML: logistic regression trained on Past_Decision; probabilities are out-of-sample (leave-one-out).",
 "- Change any blue number -> brackets (bands), DTI, rule decisions and this summary update automatically.",
 "- ML probabilities are stored values; re-run the notebook after changing data to re-train.",
 "- Data are synthetic (statistically plausible) and for learning only; with 20 rows ML results are illustrative."]
for k, t in enumerate(notes): sm.cell(11 + k, 1, t)
sm["A12"].font = Font(bold=True)

wb.active = 0   # open on the Applicants sheet
wb.save(OUTPUT_FILE)

# Optional: let LibreOffice calculate and store the formula results so every viewer shows numbers immediately
soffice = shutil.which("soffice") or shutil.which("libreoffice")
if soffice:
    os.makedirs("_recalc", exist_ok=True)
    subprocess.run([soffice, "--headless", "--convert-to", "xlsx", "--outdir", "_recalc", OUTPUT_FILE],
                   check=False, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    if os.path.exists(os.path.join("_recalc", OUTPUT_FILE)):
        shutil.move(os.path.join("_recalc", OUTPUT_FILE), OUTPUT_FILE)
print("Saved", OUTPUT_FILE)
```

## Key takeaways
* **Traditional programming**: transparent and easy to audit (every rejection has a stated reason), but it only knows the rules we thought of. It cannot adapt unless a person edits the code.
* **Machine learning**: learns patterns (including the officers' judgement calls) from data, gives a probability, and adapts when re-trained — but it needs enough good data, and with only 20 rows it is a demonstration, not a production model.
* In this tiny dataset the model gives `DTI` the biggest weight and `Credit_Score` almost none, because the low-score applicants also have high DTI. That is a small-data artefact, and a good reminder that ML learns from whatever patterns the data happens to contain.
* The cases where the two **disagree** are the interesting ones: they show where human judgement went beyond the written rules.
